```python
def room_action(is_dirty):
    if is_dirty is None:
        return "CHECK"
    elif is_dirty:
        return "SUCK"
    else:
        return "MOVE"
for value in (True, False, None):
    print(value, room_action(value))
```

```python
def reflex_agent(location, status):
    if status == "Dirty":
        return "SUCK"
    elif location == "A":
        return "RIGHT"
    elif location == "B":
        return "LEFT"
    else:
        return "STOP"
tests = [("A", "Dirty"), ("A", "Clean"), ("B", "Dirty"), ("B", "Clean")]
for location, status in tests:
    print(location, status, "->", reflex_agent(location, status))
```

```python
def reflex_agent(location, status):
    if location not in ("A", "B"):
        return "STOP"
    if status == "Dirty":
        return "SUCK"
    elif location == "A":
        return "RIGHT"
    else:
        return "LEFT"
print(reflex_agent("C", "Dirty"))
```

```python
rooms = {"A": "Dirty", "B": "Dirty"}
location = "A"
actions = []
for step in range(6):
    status = rooms[location]
    action = reflex_agent(location, status)
    actions.append(action)
    print(f"Step {step+1}: {location}, {status} -> {action}")
    if action == "SUCK":
        rooms[location] = "Clean"
    elif action == "RIGHT":
        location = "B"
    elif action == "LEFT":
        location = "A"
    else:
        break
print("Final rooms:", rooms)
print("Number of actions:", len(actions))
```

```python
def smart_agent(location, status, rooms):
    if all(value == "Clean" for value in rooms.values()):
        return "STOP"
    if status == "Dirty":
        return "SUCK"
    if location == "A":
        return "RIGHT"
    if location == "B":
        return "LEFT"
    return "STOP"
```

```python
def simulate_smart(initial_rooms):
    rooms = initial_rooms.copy()
    location = "A"
    actions = []
    for step in range(6):
        status = rooms[location]
        action = smart_agent(location, status, rooms)
        actions.append(action)
        print(f"Step {step+1}: {location}, {status} -> {action}")
        if action == "SUCK":
            rooms[location] = "Clean"
        elif action == "RIGHT":
            location = "B"
        elif action == "LEFT":
            location = "A"
        else:
            break
    print("Final rooms:", rooms)
    print("Actions:", actions)
    print("Both clean:", all(v == "Clean" for v in rooms.values()))
    return rooms, actions
for initial in ({"A": "Dirty", "B": "Dirty"}, {"A": "Clean", "B": "Dirty"}, {"A": "Clean", "B": "Clean"}):
    print("Initial rooms:", initial)
    simulate_smart(initial)
```
