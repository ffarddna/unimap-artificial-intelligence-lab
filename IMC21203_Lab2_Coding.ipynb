```python
from sklearn.datasets import load_iris
import pandas as pd
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target
print("Dataset shape:", X.shape)
print("Feature names:", list(X.columns))
print("Species:", list(iris.target_names))
print(X.head())
```

```python
def rule_predict(petal_length, petal_width):
    if petal_length < 2.5:
        return 0
    elif petal_width < 1.7:
        return 1
    else:
        return 2
example_class = rule_predict(1.4, 0.2)
print("Rule prediction:", iris.target_names[example_class])
for length, width in [(1.4, 0.2), (4.7, 1.4), (5.6, 2.1)]:
    print(length, width, iris.target_names[rule_predict(length, width)])
```

```python
def rule_predict_changed(petal_length, petal_width):
    if petal_length < 2.5:
        return 0
    elif petal_width < 1.3:
        return 1
    else:
        return 2
for length, width in [(1.4, 0.2), (4.7, 1.4), (5.6, 2.1)]:
    before = iris.target_names[rule_predict(length, width)]
    after = iris.target_names[rule_predict_changed(length, width)]
    print((length, width), "Before:", before, "After:", after, "Changed:", before != after)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Training:", len(X_train), "samples")
print("Testing:", len(X_test), "samples")
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
model = DecisionTreeClassifier(max_depth=3, random_state=42)
model.fit(X_train, y_train)
ml_predictions = model.predict(X_test)
print("ML test accuracy:", round(accuracy_score(y_test, ml_predictions), 3))
print("Confusion matrix:\n", confusion_matrix(y_test, ml_predictions))
```

```python
rule_predictions = [
    rule_predict(row.iloc[2], row.iloc[3])
    for _, row in X_test.iterrows()
]
rule_acc = accuracy_score(y_test, rule_predictions)
ml_acc = accuracy_score(y_test, ml_predictions)
print(f"Rule-based test accuracy: {rule_acc:.3f}")
print(f"ML test accuracy:         {ml_acc:.3f}")
print("Correct (rules):", sum(y_test.to_numpy() == rule_predictions))
print("Correct (ML):", sum(y_test.to_numpy() == ml_predictions))
```
