```python
def reflex_action(location, status):
    if location not in ("A", "B"):
        return "CHECK"
    if status not in ("clean", "dirty"):
        return "CHECK"
    if status == "dirty":
        return "SUCK"
    return "MOVE_RIGHT" if location == "A" else "MOVE_LEFT"
tests = [("A", "dirty"), ("A", "clean"),
         ("B", "dirty"), ("B", "clean"), ("B", "unknown")]
for location, status in tests:
    print(location, status, reflex_action(location, status))
```

```python
print("C", "dirty", reflex_action("C", "dirty"))
```

```python
def memory_action(location, status, memory):
    if location not in ("A", "B"):
        return "CHECK"
    if status not in ("clean", "dirty"):
        return "CHECK"
    memory[location] = status
    if status == "dirty":
        return "SUCK"
    if all(value == "clean" for value in memory.values()):
        return "STOP"
    return "MOVE_RIGHT" if location == "A" else "MOVE_LEFT"
memory = {"A": "unknown", "B": "unknown"}
sequence = [("A", "clean"), ("B", "dirty"), ("B", "clean")]
for location, status in sequence:
    action = memory_action(location, status, memory)
    print(location, status, dict(memory), action)
```

```python
plans = {"A": (0.90, 5), "B": (0.80, 7), "C": (1.00, 8)}
for weight in (10, 40):
    scores = {name: weight * safety - cost for name, (safety, cost) in plans.items()}
    best = max(scores, key=scores.get)
    print(weight, scores, best)
```
